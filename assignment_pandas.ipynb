```python
import pandas as pd
```

```python
# loading the dataset
data = pd.read_csv("/Users/lena/Desktop/Python-Tasks/adult.data.csv")
data.head()
```

**1. Calculate how many men and women (attribute *sex*) are represented in this dataset.**

```python
men = (data['sex'] == 'Male').sum()
women = (data['sex'] == 'Female').sum()

print("Men:", men)
print("Women:", women)
```

```python
print(data['sex'].value_counts())
```

**2. What is the average age of men (attribute *age*) across the entire dataset?**

```python
avg_male_age = data[data['sex'] == 'Male']['age'].mean()
print(f'Average age of men: {avg_male_age}')
```

**3. What is the proportion of United States citizens (*native-country* attribute)?**

```python
US_citizen_proportion = len(data[data['native-country'] == 'United-States'])*100/len(data)
print('The proportion of United States citizens:',US_citizen_proportion)
```

**4-5. Calculate the mean and standard deviation of the age of those who earn more than 50K per year (the *salary* attribute) and those who earn less than 50K per year**

```python
mean_less = data[data['salary'] == '<=50K']['age'].mean()
mean_more = data[data['salary'] == '>50K']['age'].mean()

std_less = data[data['salary'] == '<=50K']['age'].std()
std_more = data[data['salary'] == '>50K']['age'].std()

print(f'\nMean of the age of those who earn less than 50K per year: {mean_less}, std: {std_less}\nMean of the age of those who earn more than 50K per year: {mean_more}, std: {std_more}')
```

**6. Is it true that people who earn more than 50k have at least a higher education? (*education indicator – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters or Doctorate*)**

```python
higher_education = [
    'Bachelors',
    'Prof-school',
    'Assoc-acdm',
    'Assoc-voc',
    'Masters',
    'Doctorate'
]
result = data[data['salary'] == '>50K']['education'].isin(higher_education).all()

print(result)
```

**7. Find age statistics for each race (*race* attribute) and each gender. Use *groupby* and *describe*. Find the maximum age of males of the *Asian-Pac-Islander* race.**

```python
result = data.groupby(['race', 'sex'])['age'].max()
print(f"Max age of the Asian-Pac-Islander Male: {result['Asian-Pac-Islander', 'Male']}")
```

**8. Among whom is the proportion of high-earning individuals (>50K) higher: married or single men (*marital-status* indicator)? We consider those whose *marital-status* begins with *Married* (Married-civ-spouse, Married-spouse-absent, or Married-AF-spouse) to be married; the rest are considered single.**

```python
married = data[data['marital-status'].str.startswith('Married')]
single = data[~data['marital-status'].str.startswith('Married')]

married_percentage = len(married[married['salary']=='>50K'])*100/len(data[data['salary']=='>50K'])
single_percentage = len(single[single['salary']=='>50K'])*100/len(data[data['salary']=='>50K'])

print('Married percentage: ',married_percentage,'%')
print('Single percentage: ',single_percentage,'%')
```

**9. What is the maximum number of hours a person works per week (hours-per-week indicator)? How many people work that many hours, and what percentage of them earn a lot?**

```python
max_hours = data['hours-per-week'].max()

hard_workers = data[data['hours-per-week'] == max_hours]

rich_hard_workers=(hard_workers['salary']=='>50K').sum()*100/len(hard_workers)


print(f'Max hours per week: {max_hours}')
print(f'People working max hours: {len(hard_workers)}')
print(f'Percentage earning >50K: {rich_hard_workers:.1f}%')
```

**10. Calculate the average working hours (*hours-per-week*) of low and high earners (*salary*) for each country (*native-country*).**

```python
avg_work_hours_per_cntr = (data.groupby(['native-country', 'salary'])['hours-per-week'].mean().round(1).reset_index())
print(avg_work_hours_per_cntr)
```

**11.Group people into age groups *young*, *adult*, *retiree*, where:**
* *young* corresponds to 16-35 years old
* *adult* - 35-70 years old
* *retiree* - 70-100 years old

**Enter the name of the appropriate age group for each person in the new AgeGroup column.**

```python
data['AgeGroup'] = pd.cut(
    data['age'],
    bins=[16, 35, 70, 100],
    labels=['young', 'adult', 'retiree']
)

data.head()
```

**12-13. Determine the number of people earning >50K in each age group (AgeGroup column), and also output the name of the age group in which people most often earn more than 50K (>50K)**

```python
groups = data[data['salary'] == '>50K'].groupby('AgeGroup')['salary'].count()
print(groups)

print(f'Age group with most >50K earners: {groups.idxmax()}')
```

**14. Group people by employment type (the occupation column) and determine the number of people in each group. Then write a filter function, filter_func, that will return only those groups in which the average age (the age column) is 40 or less and in which all employees work more than 5 hours per week (the hours-per-week column).**

```python
occupation_count = data.groupby('occupation')['occupation'].count()
print(occupation_count)

def filter_func(group):
    cond1 = group['age'].mean() <= 40
    cond2 = (group['hours-per-week'] > 5).all()
    return cond1 and cond2

filtered = data.groupby('occupation').filter(filter_func)
print(filtered)
```
